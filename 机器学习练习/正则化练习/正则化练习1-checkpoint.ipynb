`reg`
来自 regression（回归）

指的是 回归任务：预测一个连续型数值 例如在生信场景里：

预测 Pfirrmann score（椎间盘退变等级，1–5 分）

预测基因表达量、蛋白浓度、临床指标（如血糖值）

在代码里：

```python
y_reg = np.random.randint(1, 6, size=50)
```
这里的 `y_reg` 就是模拟的 回归标签，取值范围 1–5。

📌 `clf`
来自 classification（分类）

指的是 分类任务：预测一个离散型类别 例如在生信场景里：

判断样本是 IVDD 患者 (1) 还是 健康对照 (0)

判断肿瘤分型（比如 Luminal vs Basal）

在代码里：

```python
y_clf = np.random.randint(0, 2, size=50)
```
这里的 y_clf 就是模拟的 分类标签，取值只有 0 或 1。

````python
#使用 def 版

import numpy as np

# 模拟生信数据：50个样本，100个基因
np.random.seed(42)
X = np.random.randn(50, 100)   # 表达矩阵
y_reg = np.random.randint(1, 6, size=50)   # Pfirrmann score (1-5)
y_clf = np.random.randint(0, 2, size=50)   # IVDD标签 (0/1)

# 标准化
X = (X - X.mean(axis=0)) / X.std(axis=0)

def gradient_descent(X, y, loss_fn, grad_fn, lr=0.01, lam=0.1, epochs=1000):
    m, n = X.shape
    theta = np.zeros(n)
    losses = []
    for _ in range(epochs):
        loss = loss_fn(X, y, theta, lam)
        grad = grad_fn(X, y, theta, lam)
        theta -= lr * grad
        losses.append(loss)
    return theta, losses

# 线性回归
def loss_linear(X, y, theta, lam):
    m = len(y)
    pred = X @ theta
    return (1/(2*m)) * np.sum((pred - y)**2) + (lam/(2*m)) * np.sum(theta**2)

def grad_linear(X, y, theta, lam):
    m = len(y)
    return (1/m) * (X.T @ (X @ theta - y)) + (lam/m) * theta

theta_lin, losses_lin = gradient_descent(X, y_reg, loss_linear, grad_linear)

# 逻辑回归
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def loss_logistic(X, y, theta, lam):
    m = len(y)
    pred = sigmoid(X @ theta)
    return -(1/m) * np.sum(y*np.log(pred+1e-8) + (1-y)*np.log(1-pred+1e-8)) + (lam/(2*m)) * np.sum(theta**2)

def grad_logistic(X, y, theta, lam):
    m = len(y)
    pred = sigmoid(X @ theta)
    return (1/m) * (X.T @ (pred - y)) + (lam/m) * theta

theta_log, losses_log = gradient_descent(X, y_clf, loss_logistic, grad_logistic)
````

````python
import numpy as np
import pandas as pd

# 模拟生信数据：50个样本，100个基因
np.random.seed(42)
X = np.random.randn(50,100)   # 生成二维矩阵，d0 = 50,d1 = 100,既是（50，100）,np.random.randn 不支持关键字参数 d0, d1，只能省略d0和d1
y_reg = np.random.randint(low = 1,high = 6, size=50)   # low=1, high=6 → 会生成 1 到 5 之间的整数,一共生成 50 个数,在生信场景里，这里模拟的是 Pfirrmann score（椎间盘退变等级）
y_clf = np.random.randint(0, 2, size=50)   # IVDD标签 (0/1)

# 标准化
X = (X - X.mean(axis=0)) / X.std(axis=0)


# 参数设置
m, n = X.shape        # 把行数赋给 m（样本数），列数赋给 n（特征数） 也可以 m = X.shape[0],n = X.shape[1]
theta = np.zeros(n)   # 初始化参数
lr = 0.01             # 学习率
lam = 0.1             # 正则化系数
epochs = 1000         # 迭代次数

losses = []

# 梯度下降循环
for epoch in range(epochs):
    # 预测值
    y_pred = X @ theta    # 等价于 y_pred = X.dot(theta)
    
    # 损失函数 (带正则化)
    loss = (1/(2*m)) * np.sum((y_pred - y_reg)**2) + (lam/(2*m)) * np.sum(theta**2)
    losses.append(loss)
    
    # 梯度 (带正则化)
    grad = (1/m) * (X.T @ (y_pred - y_reg)) + (lam/m) * theta
    
    # 参数更新
    theta -= lr * grad
    
    # 每100次打印一次
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")
````

换标签：y_reg → y_clf

换预测函数：y_pred = sigmoid(X @ theta)

换损失函数：交叉熵 + L2 正则化
