研究目标：

我们有 转录组测序数据（RNA-seq），每个样本测了若干基因的表达量。

我们关心的是 椎间盘退变 (IVDD) 的两个指标：

Pfirrmann score（1–5 分，连续型，表示退变程度）

IVDD 患者标签（0=健康，1=患者，二分类）

数据：

特征矩阵 
𝑋 ：每行是一个样本，每列是一个基因表达量（标准化后）。

标签：

𝑦_reg ：Pfirrmann score（回归任务）

𝑦_clf ：IVDD 标签（分类任务）

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
m, n = 100, 500
X = np.random.randn(m, n)
y_reg = np.random.randint(1, 6, size=m)
y_clf = np.random.randint(0, 2, size=m)

# 标准化
X = (X - X.mean(axis=0)) / X.std(axis=0)

# ========== 正则化线性回归 ==========
beta_reg = np.zeros(n)
lr, lam, epochs = 0.01, 0.1, 500
losses_reg = []

for epoch in range(epochs):
    y_pred = X @ beta_reg
    err = y_pred - y_reg
    loss = (1/(2*m))*np.sum(err**2) + (lam/(2*m))*np.sum(beta_reg**2)
    losses_reg.append(loss)
    grad = (1/m)*(X.T @ err) + (lam/m)*beta_reg
    beta_reg -= lr*grad

plt.plot(losses_reg, label="Linear Regression Loss")
plt.legend()
plt.show()

# ========== 正则化逻辑回归 ==========
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

beta_clf = np.zeros(n)
losses_clf = []

for epoch in range(epochs):
    p = sigmoid(X @ beta_clf)
    loss = -(1/m)*np.sum(y_clf*np.log(p+1e-8) + (1-y_clf)*np.log(1-p+1e-8)) + (lam/(2*m))*np.sum(beta_clf**2)
    losses_clf.append(loss)
    grad = (1/m)*(X.T @ (p - y_clf)) + (lam/m)*beta_clf
    beta_clf -= lr*grad

plt.plot(losses_clf, label="Logistic Regression Loss", color="red")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
m,n = 100,500
X = np.random.randn(m,n)
y_reg = np.random.randint(1, 6, size=m)
y_clf = np.random.randint(0, 2, size=m)

theta = np.zeros(n)
lam = 0.1
lr = 0.001
epochs = 3000
losses = []

for epoch in range(epochs):
    y_pred = X @ (theta)
    error = y_pred - y_reg
    loss = (1 / (2*m))*np.sum(error**2) + (lam / (2*m))*np.sum(theta**2)
    losses.append(loss)
    grad = (1 / m) * ( X.T @ (error)) + (lam / m)*(theta)
    theta -= lr*grad

plt.plot(losses, label="Ridge Loss")
plt.legend() 
plt.show()


# 岭回归解析解：
A = X.T @ X + lam * np.eye(n)
b = X.T @ y_reg
theta_star = np.linalg.solve(A, b)

resid = X @ theta_star - y_reg
mse = np.mean(resid**2)
print("最优解的均方误差:", mse)

# 计算岭回归的闭式解和残差：

m, n = 100, 500
X = np.random.randn(m, n)
y = np.random.randint(1, 6, size=m)
lam = 0.1

# 岭回归闭式解
A = X.T @ X + lam * np.eye(n)
b = X.T @ y
theta_star = np.linalg.solve(A, b)

# 残差
resid = X @ theta_star - y
mse = np.mean(resid**2)
print("Ridge MSE:", mse)


# 检查普通最小二乘解 (λ=0)，使用伪逆：

theta_ols = np.linalg.pinv(X) @ y
resid_ols = X @ theta_ols - y
mse_ols = np.mean(resid_ols**2)
print("OLS MSE:", mse_ols)

#'''如果 mse_ols 接近 0，说明存在一个精确解，可以完全拟合数据，损失为零。

# 如果 mse_ols 显著大于 0，说明数据无法完全拟合，损失函数不可能收敛到 0。
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

m,n = 100,500
X = np.random.randn(m,n)
y_reg = np.random.randint(1, 6, size=m)
y_clf = np.random.randint(0, 2, size=m)

# 根据 y 的形状来确定是 X@theta 还是 theta@X
y_clf.shape

# 所有应该是 X @ theta，推算出 theta 形状
theta_clf = np.zeros(n)


# 标准化特征矩阵
X = (X - X.mean(axis=0)) / X.std(axis=0)



# 普通逻辑回归
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

lr = 0.01
epochs = 3000
losses_clf = []
for epoch in range(epochs):
    z = X @ theta_clf
    y_pred = sigmoid(z)   # (100,)
    loss = -(1/m) * np.sum(y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8))
    losses_clf.append(loss)
    grad = (1/m) * (X.T @ (y_pred - y_clf)) # (500,)
    theta_clf -= lr * grad

plt.plot(losses_clf, label="Logistic Regression Loss")
plt.legend()
plt.show()


# 正则化逻辑回归
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
lr = 0.01
lam = 0.1
epochs = 3000
theta_clf_reg = np.zeros(n)   # 初始化参数
losses_clf_reg = []
for epoch in range(epochs):
    # 线性组合
    z = X @ theta_clf_reg   # (100,)
    
    # 预测概率
    y_pred = sigmoid(z)     # (100,)

    # 误差
    error = y_pred - y_clf    # (100,)

    # 损失函数：交叉熵 + L2 正则化 
    loss = -(1/m) * np.sum( y_clf * np.log(y_pred + 1e-8) + \
                           (1 - y_clf) * np.log(1 - y_pred + 1e-8) ) + \
                            (lam/(2*m)) * np.sum(theta_clf_reg**2) 
    losses_clf_reg.append(loss)

    # 梯度：普通逻辑回归梯度 + L2 正则化项 
    grad = (1/m) * (X.T @ (error)) + (lam/m) * theta_clf_reg
    
    # 参数更新 
    theta_clf_reg -= lr * grad


plt.plot(losses_clf, label="Logistic Regression Loss")
plt.plot(losses_clf_reg, label="Regularized Logistic Regression Loss")
plt.legend()
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

m,n = 100,500
X = np.random.randn(m,n)
y_reg = np.random.randint(1, 6, size=m)
y_clf = np.random.randint(0, 2, size=m)

# 根据 y 的形状来确定是 X@theta 还是 theta@X
y_clf.shape

# 所有应该是 X @ theta，推算出 theta 形状
theta_clf = np.zeros(n)


# 标准化特征矩阵
X = (X - X.mean(axis=0)) / X.std(axis=0)



# 普通逻辑回归
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# L1 正则化逻辑回归
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

lr = 0.01
lam = 0.1
epochs = 3000
theta_clf_l1 = np.zeros(n)   # 初始化参数
losses_clf_l1 = []

for epoch in range(epochs):
    # 线性组合
    z = X @ theta_clf_l1
    
    # 预测概率
    y_pred = sigmoid(z)
    
    # 损失函数：交叉熵 + L1 正则化
    loss = -(1/m) * np.sum(
        y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)
    ) + (lam/m) * np.sum(np.abs(theta_clf_l1))
    losses_clf_l1.append(loss)
    
    # 梯度：普通逻辑回归梯度 + L1 正则化项 (次梯度)
    grad = (1/m) * (X.T @ (y_pred - y_clf)) + (lam/m) * np.sign(theta_clf_l1)
    
    # 参数更新
    theta_clf_l1 -= lr * grad

# 绘制损失曲线
plt.plot(losses_clf_l1, label="L1 Regularized Logistic Regression Loss")
plt.legend()
plt.show()


```
