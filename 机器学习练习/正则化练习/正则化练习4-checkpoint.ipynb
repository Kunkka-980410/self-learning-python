研究目标：

我们有 转录组测序数据（RNA-seq），每个样本测了若干基因的表达量。

我们关心的是 椎间盘退变 (IVDD) 的两个指标：

Pfirrmann score（1–5 分，连续型，表示退变程度）

IVDD 患者标签（0=健康，1=患者，二分类）

数据：

特征矩阵 
𝑋 ：每行是一个样本，每列是一个基因表达量（标准化后）。

标签：

𝑦_reg ：Pfirrmann score（回归任务）

𝑦_clf ：IVDD 标签（分类任务）

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
m, n = 100, 500
X = np.random.randn(m, n)
y_reg = np.random.randint(1, 6, size=m)
y_clf = np.random.randint(0, 2, size=m)

# 标准化
X = (X - X.mean(axis=0)) / X.std(axis=0)

# ========== 正则化线性回归 ==========
beta_reg = np.zeros(n)
lr, lam, epochs = 0.01, 0.1, 500
losses_reg = []

for epoch in range(epochs):
    y_pred = X @ beta_reg
    err = y_pred - y_reg
    loss = (1/(2*m))*np.sum(err**2) + (lam/(2*m))*np.sum(beta_reg**2)
    losses_reg.append(loss)
    grad = (1/m)*(X.T @ err) + (lam/m)*beta_reg
    beta_reg -= lr*grad

plt.plot(losses_reg, label="Linear Regression Loss")
plt.legend()
plt.show()

# ========== 正则化逻辑回归 ==========
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

beta_clf = np.zeros(n)
losses_clf = []

for epoch in range(epochs):
    p = sigmoid(X @ beta_clf)
    loss = -(1/m)*np.sum(y_clf*np.log(p+1e-8) + (1-y_clf)*np.log(1-p+1e-8)) + (lam/(2*m))*np.sum(beta_clf**2)
    losses_clf.append(loss)
    grad = (1/m)*(X.T @ (p - y_clf)) + (lam/m)*beta_clf
    beta_clf -= lr*grad

plt.plot(losses_clf, label="Logistic Regression Loss", color="red")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 数据准备
m, n = 100, 500
X = np.random.randn(m, n)
y_clf = np.random.randint(0, 2, size=m)

# 标准化特征矩阵
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Sigmoid函数
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 超参数
lr = 0.01
lam = 0.1
epochs = 3000

# ---------------- 普通逻辑回归 ----------------
theta_clf = np.zeros(n)
losses_clf = []
for epoch in range(epochs):
    z = X @ theta_clf
    y_pred = sigmoid(z)
    loss = -(1/m) * np.sum(y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8))
    losses_clf.append(loss)
    grad = (1/m) * (X.T @ (y_pred - y_clf))
    theta_clf -= lr * grad

# ---------------- L2 正则化逻辑回归 ----------------
theta_clf_reg = np.zeros(n)
losses_clf_reg = []
for epoch in range(epochs):
    z = X @ theta_clf_reg
    y_pred = sigmoid(z)
    error = y_pred - y_clf
    loss = -(1/m) * np.sum(y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)) \
           + (lam/(2*m)) * np.sum(theta_clf_reg**2)
    losses_clf_reg.append(loss)
    grad = (1/m) * (X.T @ error) + (lam/m) * theta_clf_reg
    theta_clf_reg -= lr * grad

# ---------------- L1 正则化逻辑回归（惩罚截距） ----------------
theta_clf_l1 = np.zeros(n)
losses_clf_l1 = []
for epoch in range(epochs):
    z = X @ theta_clf_l1
    y_pred = sigmoid(z)
    loss = -(1/m) * np.sum(y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)) \
           + (lam/m) * np.sum(np.abs(theta_clf_l1))
    losses_clf_l1.append(loss)
    grad = (1/m) * (X.T @ (y_pred - y_clf)) + (lam/m) * np.sign(theta_clf_l1)
    theta_clf_l1 -= lr * grad

# ---------------- L1 正则化逻辑回归（不惩罚截距） ----------------
theta_clf_l1_nobias = np.zeros(n)
losses_clf_l1_nobias = []
for epoch in range(epochs):
    z = X @ theta_clf_l1_nobias
    y_pred = sigmoid(z)
    loss = -(1/m) * np.sum(y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)) \
           + (lam/m) * np.sum(np.abs(theta_clf_l1_nobias[1:]))   # 跳过截距
    losses_clf_l1_nobias.append(loss)
    grad = (1/m) * (X.T @ (y_pred - y_clf))
    grad[1:] += (lam/m) * np.sign(theta_clf_l1_nobias[1:])       # 不惩罚 theta[0]
    theta_clf_l1_nobias -= lr * grad

# ---------------- 绘制对比图 ----------------
plt.figure(figsize=(10,6))
plt.plot(losses_clf, label="Logistic Regression (no regularization)")
plt.plot(losses_clf_reg, label="L2 Regularized Logistic Regression")
plt.plot(losses_clf_l1, label="L1 Regularized Logistic Regression (with bias penalty)")
plt.plot(losses_clf_l1_nobias, label="L1 Regularized Logistic Regression (no bias penalty)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Comparison of Logistic Regression Variants")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 数据准备
m, n = 100, 500
X = np.random.randn(m, n)
y_clf = np.random.randint(0, 2, size=m)

# 标准化特征矩阵
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Sigmoid函数
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 超参数
lr = 0.01
lam = 0.1
alpha = 0.5   # L1 与 L2 混合比例
epochs = 3000

# ---------------- Elastic Net 正则化逻辑回归（惩罚截距） ----------------
theta_enet_bias = np.zeros(n)
losses_enet_bias = []

for epoch in range(epochs):
    z = X @ theta_enet_bias
    y_pred = sigmoid(z)
    
    # 损失函数：交叉熵 + Elastic Net 正则化（惩罚所有参数）
    loss = -(1/m) * np.sum(
        y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)
    ) + alpha * (lam/m) * np.sum(np.abs(theta_enet_bias)) \
      + (1-alpha) * (lam/(2*m)) * np.sum(theta_enet_bias**2)
    losses_enet_bias.append(loss)
    
    # 梯度
    grad = (1/m) * (X.T @ (y_pred - y_clf))
    grad += alpha * (lam/m) * np.sign(theta_enet_bias) \
          + (1-alpha) * (lam/m) * theta_enet_bias
    
    theta_enet_bias -= lr * grad

# ---------------- Elastic Net 正则化逻辑回归（不惩罚截距） ----------------
theta_enet_nobias = np.zeros(n)
losses_enet_nobias = []

for epoch in range(epochs):
    z = X @ theta_enet_nobias
    y_pred = sigmoid(z)
    
    # 损失函数：交叉熵 + Elastic Net 正则化（跳过截距）
    loss = -(1/m) * np.sum(
        y_clf * np.log(y_pred + 1e-8) + (1 - y_clf) * np.log(1 - y_pred + 1e-8)
    ) + alpha * (lam/m) * np.sum(np.abs(theta_enet_nobias[1:])) \
      + (1-alpha) * (lam/(2*m)) * np.sum(theta_enet_nobias[1:]**2)
    losses_enet_nobias.append(loss)
    
    # 梯度
    grad = (1/m) * (X.T @ (y_pred - y_clf))
    grad[1:] += alpha * (lam/m) * np.sign(theta_enet_nobias[1:]) \
              + (1-alpha) * (lam/m) * theta_enet_nobias[1:]
    
    theta_enet_nobias -= lr * grad

# ---------------- 绘制对比图 ----------------
plt.figure(figsize=(10,6))
plt.plot(losses_enet_bias, label="Elastic Net (with bias penalty)")
plt.plot(losses_enet_nobias, label="Elastic Net (no bias penalty)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Elastic Net Logistic Regression Comparison")
plt.legend()
plt.show()
```

```python
import numpy as np

np.random.seed(42)  # 保证结果可复现

# 1. 构建模拟数据
n_samples = 300   # 样本数
n_genes = 20      # 基因数

# 基因表达矩阵 (N x 20)，每个基因表达量服从标准正态分布
X = np.random.normal(0, 1, (n_samples, n_genes))

# 2. 连续变量：肿瘤大小（由基因 1-5 决定）
# 给基因 1-5 设置不同的权重，并加上噪声
tumor_size = (
    X[:,0]*2 +
    X[:,1]*1.5 +
    X[:,2]*(-1) +
    X[:,3]*0.5 +
    X[:,4]*3 +
    np.random.normal(0, 2, n_samples)  # 加噪声
)

# 3. 分类变量：肿瘤是否恶性（由基因 6-10 决定）
# 先构造 logits，再通过 sigmoid 转换为概率
logits = (
    X[:,5]*2 +
    X[:,6]*(-1.5) +
    X[:,7]*1 +
    X[:,8]*2.5 +
    X[:,9]*(-2)
)

prob = 1 / (1 + np.exp(-logits))  # sigmoid
y_class = (prob > 0.5).astype(int)  # 二分类标签

# 4. 整理目标变量形状
y_reg = tumor_size.reshape(-1, 1)  # 线性回归目标 (N,1)
y_clf = y_class.reshape(-1, 1)     # 逻辑回归目标 (N,1)

# 5. 标准化输入特征，并添加截距列
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_scaled = (X - X_mean) / X_std

X_lin = np.hstack([np.ones((n_samples,1)), X_scaled])  # 线性回归设计矩阵
X_log = np.hstack([np.ones((n_samples,1)), X_scaled])  # 逻辑回归设计矩阵

print("X_lin shape:", X_lin.shape)
print("y_reg shape:", y_reg.shape)
print("X_log shape:", X_log.shape)
print("y_clf shape:", y_clf.shape)

import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def train_model(X, y, loss_type="linear", lam=0.1, alpha=0.5, lr=0.01, epochs=3000):
    m, n = X.shape
    theta = np.zeros((n,1))
    losses = []
    
    for epoch in range(epochs):
        z = X @ theta
        if loss_type in ["logistic", "lasso", "elasticnet"]:
            y_hat = sigmoid(z)
            error = y_hat - y
        else:
            y_hat = z
            error = y_hat - y
        
        # 损失函数和梯度
        if loss_type == "linear":
            loss = (1/(2*m)) * np.sum(error**2)
            grad = (1/m) * (X.T @ error)
        
        elif loss_type == "ridge":
            loss = (1/(2*m)) * np.sum(error**2) + (lam/(2*m)) * np.sum(theta[1:]**2)
            grad = (1/m) * (X.T @ error)
            grad[1:] += (lam/m) * theta[1:]
        
        elif loss_type == "logistic":
            loss = -(1/m) * np.sum(y*np.log(y_hat+1e-8) + (1-y)*np.log(1-y_hat+1e-8))
            grad = (1/m) * (X.T @ error)
        
        elif loss_type == "lasso":
            loss = -(1/m) * np.sum(y*np.log(y_hat+1e-8) + (1-y)*np.log(1-y_hat+1e-8)) \
                   + (lam/m) * np.sum(np.abs(theta[1:]))
            grad = (1/m) * (X.T @ error)
            grad[1:] += (lam/m) * np.sign(theta[1:])
        
        elif loss_type == "elasticnet":
            loss = -(1/m) * np.sum(y*np.log(y_hat+1e-8) + (1-y)*np.log(1-y_hat+1e-8)) \
                   + alpha*(lam/m)*np.sum(np.abs(theta[1:])) \
                   + (1-alpha)*(lam/(2*m))*np.sum(theta[1:]**2)
            grad = (1/m) * (X.T @ error)
            grad[1:] += alpha*(lam/m)*np.sign(theta[1:]) + (1-alpha)*(lam/m)*theta[1:]
        
        losses.append(loss)
        theta -= lr * grad
    
    return theta, losses

# 示例：训练普通线性回归
theta_lin, losses_lin = train_model(X_lin, y_reg, loss_type="linear")
plt.plot(losses_lin); plt.title("Linear Regression Loss"); plt.show()

# 示例：训练逻辑回归
theta_log, losses_log = train_model(X_log, y_clf, loss_type="logistic")
plt.plot(losses_log); plt.title("Logistic Regression Loss"); plt.show()

# 示例：训练 Ridge 回归
theta_ridge, losses_ridge = train_model(X_lin, y_reg, loss_type="ridge", lam=0.1)
plt.plot(losses_ridge); plt.title("Ridge Regression Loss"); plt.show()

# 示例：训练 Lasso Logistic 回归
theta_lasso, losses_lasso = train_model(X_log, y_clf, loss_type="lasso", lam=0.1)
plt.plot(losses_lasso); plt.title("Lasso Logistic Regression Loss"); plt.show()

# 示例：训练 Elastic Net Logistic 回归
theta_en, losses_en = train_model(X_log, y_clf, loss_type="elasticnet", lam=0.1, alpha=0.5)
plt.plot(losses_en); plt.title("Elastic Net Logistic Regression Loss"); plt.show()
```
