定义函数 analyze_single_batch(batch_data):

    初始化：
        samples = {}
        quality_summary = {}
        worst_sample = None
        worst_avg = 正无穷
        均值列表 = []

    遍历每个样本：
        获取 sample_name 和 values
        计算 mean 和 std
        判断质量等级（excellent / high / medium / low）
        更新 worst_sample（如果 mean 更低）
        将样本信息存入 samples[sample_name]
        将 mean 添加到均值列表（用于排名）
        将 mean 和 std 添加到 quality_summary[quality] 的列表中

    排名处理：
        按均值从高到低排序所有样本
        给每个样本分配 rank 值

    分组统计处理：
        遍历每个质量等级：
            计算 avg_mean 和 avg_std
            存入 quality_summary[quality]

    返回结构：
        {
            "samples": samples,
            "quality_summary": quality_summary,
            "worst_sample": worst_sample
        }


````python
batches = {
    "Batch_1": {
        "Sample_A": [30, 32, 35, 40],
        "Sample_B": [25, 28, 22, 29],
        "Sample_C": [15, 18, 20, 22],
        "Sample_D": [40, 42, 45, 50],
        "Sample_E": [20, 21, 22, 23]
    },
    "Batch_2": {
        "Sample_F": [10, 12, 8, 14],
        "Sample_G": [38, 36, 39, 40],
        "Sample_H": [20, 21, 22, 23],
        "Sample_I": [30, 31, 32, 33],
        "Sample_J": [18, 19, 20, 21]
    },
    "Batch_3": {
        "Sample_K": [45, 46, 47, 48],
        "Sample_L": [28, 29, 30, 31],
        "Sample_M": [19, 20, 21, 22],
        "Sample_N": [35, 36, 37, 38],
        "Sample_O": [15, 16, 17, 18]
    }
}


batch_info = {}
Sample_info = {}

for batch,sample_info in batches.items():
    quality_summary = {}
    
    for Sample_name,values in sample_info.items():
        mean = sum(values)/len(values)
        
        if all(v >= 40 for v in values):
            quality = 'excellent'
        elif all(v >= 30 for v in values):
            quality = 'high'
        elif all(v >= 20 for v in values):
            quality = 'medium'
        else:
            quality = 'low'

 # 正确计算标准差
        square_diffs = [(v - mean) ** 2 for v in values]   #列表化
        std = round(((sum(square_diffs) / len(values)))**0.5, 2)
            

            
        Sample_info[Sample_name] = {'quality': quality,
            'mean': mean,
            'values': values,
            'std':std
        }
        # 初始化该质量等级的统计容器
        if quality not in quality_summary:
            quality_summary[quality] = {
                "count": 0,
                "total_mean": 0,
                "total_std": 0
            }

        # 累加统计数据
        quality_summary[quality]["count"] += 1
        quality_summary[quality]["total_mean"] += mean
        quality_summary[quality]["total_std"] += std

    # 计算每个质量等级的平均值
    for q in quality_summary:
        count = quality_summary[q]["count"]
        avg_mean = round(quality_summary[q]["total_mean"] / count, 2)
        avg_std = round(quality_summary[q]["total_std"] / count, 2)

        # 替换为最终结构
        quality_summary[q] = {
            "count": count,
            "avg_mean": avg_mean,
            "avg_std": avg_std
        }
        
mean_list = []
for name in Sample_info:
    mean = Sample_info[name]["mean"]
    pair = [name, mean]
    mean_list.append(pair)

# 冒泡排序（从高到低）
for i in range(len(mean_list)):
    for j in range(len(mean_list) - i - 1):
        if mean_list[j][1] < mean_list[j + 1][1]:
            # 交换位置
            temp = mean_list[j]
            mean_list[j] = mean_list[j + 1]
            mean_list[j + 1] = temp

'''
def get_mean(pair):
    return pair[1]

sorted_list = sorted(mean_list, key=get_mean, reverse=True)

'''
# 分配排名
rank_dict = {}
rank = 1
for pair in mean_list:
    name = pair[0]
    rank_dict[name] = rank
    rank += 1

# 写入字典
for name in Sample_info:
    Sample_info[name]["rank"] = rank_dict[name]


print(Sample_info)
````

这两段代码在执行效果上是完全一样的，但在结构逻辑上略有不同。我们来拆解一下：

✅ 第一段代码：
```python
if q not in quality_summary:
    quality_summary[q] = {'count': 0, 'mean_list': [], 'std_list': []}
else:
    quality_summary[q]['count'] += 1
    quality_summary[q]['mean_list'].append(mean)
```

初始化和更新是分开的。

如果 q 是第一次出现，只初始化，不做统计。

统计逻辑只在 else 中执行。

