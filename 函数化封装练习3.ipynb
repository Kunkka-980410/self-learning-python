````python
def classify_samples(batches):
    sample_quality = {}
    for batch, sample in batches.items():
        for key, value in sample.items():
            if all(i >= 30 for i in value):
                sample_quality[key] = 'high'
            elif all(i >= 20 for i in value) and any(i < 30 for i in value):
                sample_quality[key] = 'medium'
            elif any(i < 20 for i in value):
                sample_quality[key] = 'low'
    return sample_quality


def count_batch_quality(batches, sample_quality):
    batch_quality_count = {}
    for batch, samples in batches.items():
        batch_quality_count[batch] = {"high": 0, "medium": 0, "low": 0}
        for sample_name in samples:
            quality = sample_quality[sample_name]
            batch_quality_count[batch][quality] += 1
    return batch_quality_count
````

第二次循环：batch = "Batch_2"，sample = {"Sample_F": [...], "Sample_G": [...], ...}

`for sample_name, values in sample.items()`

这是在遍历当前批次下的每个样本：

比如在 "Batch_1" 中：

第一次：sample_name = "Sample_A"，values = [30, 32, 35, 40]

第二次：sample_name = "Sample_B"，values = [25, 28, 22, 29]

def batch_quality_averages_and_worst(batches):
    def classify(values):
        if all(v >= 40 for v in values):
            return "excellent"
        elif all(v >= 30 for v in values):
            return "high"
        elif all(v >= 20 for v in values):
            return "medium"
        else:
            return "low"

    result = {}

    for batch_name, samples in batches.items():
        quality_groups = {}
        worst_sample = None
        worst_score = float("inf")

        for sample_name, values in samples.items():
            quality = classify(values)
            if quality not in quality_groups:
                quality_groups[quality] = []
            quality_groups[quality].extend(values)

            total_score = sum(values)
            if total_score < worst_score:
                worst_score = total_score
                worst_sample = sample_name

        # 计算每个质量等级的平均值
        averages = {
            q: round(sum(scores) / len(scores), 2)
            for q, scores in quality_groups.items()
        }

        result[batch_name] = {
            "averages": averages,
            "worst_sample": worst_sample
        }

    return result

````python
batch_quality_avg = {}

for batch_name, sample_group in batches.items():
    quality_scores = {}
    worst_sample = None
    worst_avg = float("inf")

    for sample_name, values in sample_group.items():
        # 分类
        if all(v >= 40 for v in values):
            quality = 'excellent'
        elif all(v >= 30 for v in values):
            quality = 'high'
        elif all(v >= 20 for v in values):
            quality = 'medium'
        else:
            quality = 'low'

        # 分组收集
        if quality not in quality_scores:
            quality_scores[quality] = []
        quality_scores[quality].extend(values)

        # 最差样本判断
        avg_score = sum(values) / len(values)
        if avg_score < worst_avg:
            worst_avg = avg_score
            worst_sample = sample_name

    # 计算均值
    quality_avg = {
        q: round(sum(scores) / len(scores), 2)
        for q, scores in quality_scores.items()
    }

    # 加入最差样本
    quality_avg["worst_sample"] = worst_sample
    batch_quality_avg[batch_name] = quality_avg


for batch, info in batch_quality_avg.items():
    print(f"{batch}:")
    for k, v in info.items():
        print(f"  {k}: {v}")


# 🧠 为什么用 float("inf") 而不是 0？
# 因为我们要找的是“最小值”，所以初始值必须比所有可能的均值都大。否则你可能会误判，比如如果你用 worst_avg = 0，那么所有样本均值都比它大，永远不会更新。
````
