````python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    }
}
````

````python
batch_info = {}
for batch,sample_info in rna_batches.items():
    Samples = {}
    for sample_name,values in sample_info.items():
        if all(v >= 30 for v in values):
            quality = 'excellent'
        elif all(v >= 20 for v in values):
            quality = 'high' 
        elif all(v >= 10 for v in values):
            quality = 'medium'
        elif all(v >= 5 for v in values):
            quality = 'low'
        else:
            quality = 'bad'
        
        mean = sum(values)/len(values)
        var = [(v - mean)**2 for v in values]
        var_mean = sum(var)/len(var)
        std = var_mean**0.5
        Samples[sample_name] ={'quality':quality,'mean':mean,'std':std}
    mean_list = []
    for sample_name,details in Samples.items():
        mean_list.append((sample_name,details['mean']))
    sorted_mean = sorted(mean_list,key=lambda x:x[1],reverse=True )
    rank = 1
    for rank,(name,_) in enumerate(sorted_mean,start=1):
        Samples[name]['rank'] = rank
        rank += 1
    batch_info[batch] = Samples
print(batch_info)
````

````python
# 构造 mean_list
mean_list = []
for Sample_name, details in Samples.items():
    mean_list.append((Sample_name, details['mean']))

# 冒泡排序：按均值从高到低排序
for i in range(len(mean_list)):
    for j in range(len(mean_list) - i - 1):
        if mean_list[j][1] < mean_list[j + 1][1]:
            # 交换位置
            mean_list[j], mean_list[j + 1] = mean_list[j + 1], mean_list[j]

# 分配排名
rank = 1
for Sample_name, _ in mean_list:
    Samples[Sample_name]['rank'] = rank
    rank += 1
````

✅ 逐行解释
🔹 第 1 行：
```python
for i in range(len(mean_list)):
```
外层循环控制“轮数”，也就是你要进行多少次“冒泡”。

每一轮都会把当前未排序部分中最大的值“冒”到最右边。

`len(mean_list)` 是总样本数，最多需要这么多轮。

🔹 第 2 行：
```python
for j in range(len(mean_list) - i - 1):
```
内层循环控制每一轮的“相邻比较”次数。

每次比较 `j` 和 `j+1` 两个位置的值。

`- i - 1` 是因为每轮结束后，最后 i 个元素已经排好序，不需要再比较。

🔹 第 3 行：
```python
if mean_list[j][1] < mean_list[j + 1][1]:
```
比较两个相邻样本的均值（[1] 是均值）。

如果左边的样本均值比右边小，就交换位置（因为我们要降序排序）。

🔹 第 4 行：
```python
mean_list[j], mean_list[j + 1] = mean_list[j + 1], mean_list[j]
```
交换两个样本的位置。

把较大的均值“冒”到右边。

🧪 举个例子
假设 `mean_list = [('A', 30), ('B', 40), ('C', 35)]`

第 1 轮：
比较 `A` 和 `B` → `30 < 40` → 交换 → `['B', 'A', 'C']`

比较 `A` 和 `C` → `30 < 35` → 交换 → `['B', 'C', 'A']`

第 2 轮：
比较 `B` 和 `C` → `40 > 35` → 不交换 → `['B', 'C', 'A']`

排序完成 ✅

✅ 总结一句话
冒泡排序就是不断比较相邻元素，把最大值“冒”到右边，重复多轮直到有序。

````python
batches = {
    "Batch_1": {
        "Sample_A": [30, 32, 35, 40],
        "Sample_B": [25, 28, 22, 29],
        "Sample_C": [15, 18, 20, 22]
    },
    "Batch_2": {
        "Sample_D": [40, 42, 45, 50],
        "Sample_E": [20, 21, 22, 23],
        "Sample_F": [10, 12, 8, 14]
    },
    "Batch_3": {
        "Sample_G": [38, 36, 39, 40],
        "Sample_H": [20, 21, 22, 23],
        "Sample_I": [30, 31, 32, 33]
    },
    "Batch_4": {
        "Sample_J": [18, 19, 20, 21],
        "Sample_K": [45, 46, 47, 48],
        "Sample_L": [28, 29, 30, 31]
    },
    "Batch_5": {
        "Sample_M": [19, 20, 21, 22],
        "Sample_N": [35, 36, 37, 38],
        "Sample_O": [15, 16, 17, 18]
    }
}
# 遍历所有样本，计算标准差（保留两位小数）。

# 构造 std_list = [(样本名, 标准差)]

# 用冒泡排序将 std_list 按标准差从高到低排序。

# 给每个样本分配 std_rank，表示标准差排名（1 为最高）。
rna_batches = {}
for batch,sample_info in batches.items():
    Samples = {}
    std_list = []
    for sample_name,values in sample_info.items():
        mean = sum(values)/len(values)
        square_diff = [(v - mean)**2 for v in values]
        var = sum(square_diff)/len(square_diff)
        std = var**0.5
        
        std_list.append((sample_name,std))
        if all(v >= 40 for v in values):
            quality = 'excellent'
        elif all(v >= 30 for v in values):
            quality = 'high'
        elif all(v >= 20 for v in values):
            quality = 'medium'
        else:
            quality = 'low'  
        Samples[sample_name] = {'quality':quality,'std':std}
    for i in range(len(std_list)):
        for j in range(len(std_list)-i-1):
            if std_list[j][1]<std_list[j+1][1]:
                std_list[j],std_list[j+1] = std_list[j+1],std_list[j]
    rank = 1
    for sample_name,_ in std_list:
        Samples[sample_name]['rank'] = rank
        rank += 1

    rna_batches[batch] = Samples
print(rna_batches)
````
