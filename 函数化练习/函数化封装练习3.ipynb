batches：一个嵌套字典，结构如下：

```python
batches = {
    "Batch_1": {
        "Sample_A": [30, 28, 35, 40],
        "Sample_B": [15, 18, 20, 22]
    },
    "Batch_2": {
        "Sample_C": [10, 12, 8, 14],
        "Sample_D": [38, 36, 39, 40]
    }
}
```
threshold：一个整数，表示低质量碱基的判定阈值

处理逻辑：
使用嵌套 for 循环遍历每个批次和其中的样本

对每个样本，统计低于 threshold 的碱基比例

对每个批次，计算该批次中所有样本的平均低质量比例

返回一个字典，结构如下：

```python
{
  "Batch_1": {
    "average_low_quality": 0.25,
    "samples": {
      "Sample_A": 0.0,
      "Sample_B": 0.5
    }
  },
  "Batch_2": {
    "average_low_quality": 0.5,
    "samples": {
      "Sample_C": 1.0,
      "Sample_D": 0.0
    }
  }
}
```

````python
batches = {
    "Batch_1": {
        "Sample_A": [30, 28, 35, 40],
        "Sample_B": [15, 18, 20, 22]
    },
    "Batch_2": {
        "Sample_C": [10, 12, 8, 14],
        "Sample_D": [38, 36, 39, 40]
    }
}


result = {}
threshold = 20

for batch,sample in batches.items():
    Samples ={}
    for k,v in sample.items():
        quality_scores = len(v)
        
        low_quality = []
        for i in v:
            if i < threshold:
                low_quality.append(i)
        group_quality_scores = len(low_quality)/ quality_scores
        Samples[k] = group_quality_scores



    low_quality_value = []
    for k,v in Samples.items():
        low_quality_value.append(v)
    average_low_value = sum(low_quality_value)/len(low_quality_value)
    
    
    result[batch]={
            "average_low_quality": round(average_low_value, 3),
            "samples": Samples
    }
result           
````

你有一个批次字典 batches，每个批次包含多个样本，每个样本是一组质量分数。你想根据质量分数将每个样本分为三类：

"high"：所有分数都 ≥ 30

"medium"：所有分数都 ≥ 20 且至少一个 < 30

"low"：至少一个分数 < 20

然后你要统计每个批次中这三类样本的数量。

🧠 提示：
用 all() 和 any() 判断分数范围

每个批次都初始化一个计数器字典

函数返回一个总结果字典

````python
batches = {
    "Batch_1": {
        "Sample_A": [30, 32, 35, 40],
        "Sample_B": [25, 28, 22, 29],
        "Sample_C": [15, 18, 20, 22]
    },
    "Batch_2": {
        "Sample_D": [10, 12, 8, 14],
        "Sample_E": [38, 36, 39, 40],
        "Sample_F": [20, 21, 22, 23]
    }
}

sample_quality = {}

for batch, sample in batches.items():
    for key, value in sample.items():
        # 判断是否所有值都 >= 30
        for v in value:
            if v < 30:
                break
        else:
            sample_quality[key] = 'high'
            continue  # 已分类，跳过后续判断

        # 判断是否所有值都 >= 20 且至少一个 < 30
        for v in value:
            if v < 20:
                break
        else:
            sample_quality[key] = 'medium'
            continue

        # 判断是否有任何值 < 20
        for v in value:
            if v < 20:
                sample_quality[key] = 'low'
                break      
            
````

````python
sample_quality = {}
for batch, sample in batches.items():
    for key, value in sample.items():
        if all(i >= 30 for i in value):
            sample_quality[key] = 'high'
        elif all(i >= 20 for i in value) and any(i < 30 for i in value):
            sample_quality[key] = 'medium'
        elif any(i < 20 for i in value):
            sample_quality[key] = 'low'

print(sample_quality)            
print('\n'*6)

batch_quality_count = {}

for batch, samples in batches.items():
    batch_quality_count[batch] = {"high": 0, "medium": 0, "low": 0}
    
    for sample_name in samples:
        quality = sample_quality[sample_name]
        batch_quality_count[batch][quality] += 1
print(batch_quality_count)
````
