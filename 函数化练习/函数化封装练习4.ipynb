````python
# ✅ 封装函数：analyze_batches(batches) + query_batch(batch_info, batch, key=None)

def analyze_batches(batches):
    batch_info = {}

    for batch, sample_group in batches.items():
        Sample = {}
        quality_summary = {}

        for sample_name, values in sample_group.items():
            mean = sum(values) / len(values)
            std = round((sum((v - mean) ** 2 for v in values) / len(values)) ** 0.5, 2)

            if all(v >= 40 for v in values):
                quality = 'excellent'
            elif all(v >= 30 for v in values):
                quality = 'high'
            elif all(v >= 20 for v in values):
                quality = 'medium'
            else:
                quality = 'low'

            Sample[sample_name] = {
                'values': values,
                'mean': round(mean, 2),
                'std': std,
                'quality': quality
            }

            if quality not in quality_summary:
                quality_summary[quality] = {
                    'count': 0,
                    'mean_list': [],
                    'std_list': []
                }

            quality_summary[quality]['count'] += 1
            quality_summary[quality]['mean_list'].append(mean)
            quality_summary[quality]['std_list'].append(std)

        # 排序并分配排名
        sorted_samples = sorted(Sample.items(), key=lambda x: x[1]['mean'], reverse=True)
        for rank, (sample_name, _) in enumerate(sorted_samples, start=1):
            Sample[sample_name]['rank'] = rank

        # 计算质量统计平均值
        for q in quality_summary:
            count = quality_summary[q]['count']
            mean_list_q = quality_summary[q]['mean_list']
            std_list_q = quality_summary[q]['std_list']
            avg_mean = round(sum(mean_list_q) / count, 2)
            avg_std = round(sum(std_list_q) / count, 2)

            quality_summary[q] = {
                'count': count,
                'avg_mean': avg_mean,
                'avg_std': avg_std,
                'mean_list': mean_list_q,
                'std_list': std_list_q
            }

        worst_sample = min(Sample.items(), key=lambda x: x[1]['mean'])[0]

        batch_info[batch] = {
            'sample': Sample,
            'quality_summary': quality_summary,
            'worst_sample': worst_sample
        }

    return batch_info


#✅ 查询函数：query_batch(batch_info, batch, key=None)

def query_batch(batch_info, batch, key=None):
    if batch not in batch_info:
        return f"Batch '{batch}' not found."

    if key is None:
        return batch_info[batch]

    if key == "quality_summary":
        return batch_info[batch]["quality_summary"]

    if key == "worst_sample":
        return batch_info[batch]["worst_sample"]

    if key in batch_info[batch]["sample"]:
        return batch_info[batch]["sample"][key]

    return f"Key '{key}' not found in batch '{batch}'."


#✅ 使用示例：
result = analyze_batches(batches)

query_batch(result, "Batch_1")                      # 获取整个 Batch_1 的信息
query_batch(result, "Batch_2", "Sample_G")          # 获取 Sample_G 的信息
query_batch(result, "Batch_3", "quality_summary")   # 获取 Batch_3 的质量统计
query_batch(result, "Batch_1", "worst_sample")      # 获取 Batch_1 的最差样本名
````
