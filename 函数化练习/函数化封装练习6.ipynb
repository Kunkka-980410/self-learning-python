🧬 生信练习题：RNA-seq 样本表达量分析与分组统计
你有一个 RNA-seq 表达量数据结构 rna_batches，表示多个批次中不同样本的基因表达量（TPM）。请你编写程序完成以下任务：

```python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    }
}
```

---
🎯 要求：
计算每个样本的平均表达量（mean）和标准差（std）

根据平均表达量分组：

≥ 25："high_expression"

≥ 15："medium_expression"

≥ 5："low_expression"

否则："very_low"

在每个 batch 内按平均表达量从高到低排序并分配排名

统计每个 batch 中各表达等级的样本数量和平均表达量

找出每个 batch 中表达量最低的样本

---
✅ 输出结构示例：
```python
rna_info = {
    "Batch_1": {
        "samples": {
            "Sample_A": {"values": [...], "mean": ..., "std": ..., "group": ..., "rank": ...},
            ...
        },
        "group_summary": {
            "high_expression": {"count": ..., "avg_mean": ...},
            "low_expression": {...},
            ...
        },
        "lowest_sample": "Sample_B"
    },
    ...
}
```

````python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    }
}
rna_batch_info= {}
for batch,sample_info in rna_batches.items():
    Samples = {}
    group_summary = {}
    mean_list = []
    for sample_name,values in sample_info.items():
        mean = sum(values)/len(values)
        mean_list.append((sample_name,mean))
        if all(v >= 25 for v in values):
            quality = "high_expression"
        elif all(v >= 15 for v in values):
            quality = "medium_expression"
        elif all(v >= 10 for v in values):
            quality = "low_expression"
        else:
            quality = "very low"
        Samples[sample_name] = {'quality':quality,'mean':mean}
    
    for i in range(len(mean_list)):
        for j in range(len(mean_list)-i-1):
            if mean_list[j][1] < mean_list[j+1][1]:
                mean_list[j],mean_list[j+1] = mean_list[j+1],mean_list[j]
    rank = 1
    for sample_name,_ in mean_list:
        Samples[sample_name]['rank'] = rank
        rank += 1
    
    for sample_name,details in Samples.items():
        q = details['quality']
        mean = details['mean']
        if q not in group_summary:
            group_summary[q] = {'counts':0,'mean':mean}
        group_summary[q]['counts'] += 1
        avg_mean = mean/group_summary[q]['counts']
        group_summary[q]['avg_mean']=avg_mean
    rna_batch_info[batch]={'Samples':Samples,'group_summary':group_summary}
print(Samples) 
print(mean_list)
print(group_summary)
print(rna_batch_info)
````

````python
def analyze_rna_batches(rna_batches):
    batch_info = {}
    for batch,sample_info in rna_batches.items():
        Samples = {}
        for Sample_name,values in sample_info.items():
            mean = sum(values)/len(values)
            diff = [(v - mean)**2 for v in values]
            std = (sum(diff)/len(values))**0.5
            if std < 0.3:
                label = 'stable'
            elif 0.3 <= std < 1.0:
                label = 'moderate'
            else:
                label = 'unstable'
            Samples[Sample_name] = {'std':std,'label':label}
            std_list = []
            for Sample_name,details in Samples.items():
                std = details['std']
                std_list.append((Sample_name,std))
            for i in range(len(std_list)):
                for j in range(len(std_list)-i-1):
                    if std_list[j][1] < std_list[j+1][1]:
                        std_list[j],std_list[j+1] = std_list[j+1],std_list[j]
            rank = 1
            for Sample_name,_ in std_list:
                Samples[Sample_name]['rank'] = rank
                rank += 1
            batch_info['batch'] = {'Samples':Samples}
        return batch_info
        
````

````python
'''
🧠 任务目标：
计算全局均值：每个 batch 的所有表达值的平均值

计算每个样本的偏差值：样本均值与全局均值的绝对差值

根据偏差值打标签：

偏差 < 2 → "aligned"

2 ≤ 偏差 < 5 → "mild_shift"

偏差 ≥ 5 → "strong_shift"

按偏差值降序排序并赋 rank

构造分组统计：每类标签的样本数和平均偏差

封装为函数：analyze_expression_deviation(batch_name, rna_batches)
'''

rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    },
    "Batch_3": {
        "Sample_I": [25.0, 24.5, 25.5, 26.0],
        "Sample_J": [14.0, 13.5, 14.2, 13.8],
        "Sample_K": [6.0, 6.5, 6.2, 6.1],
        "Sample_L": [19.5, 20.0, 19.8, 20.2]
    },
    "Batch_4": {
        "Sample_M": [11.0, 11.5, 11.2, 11.3],
        "Sample_N": [3.0, 3.5, 3.2, 3.1],
        "Sample_O": [16.0, 16.5, 16.2, 16.1],
        "Sample_P": [27.0, 28.5, 27.8, 28.2]
    }
}

batch_info = {}

for batch, Samples_info in rna_batches.items():
    Samples = {}
    mean_list = []

    # Step 1: 计算每个样本均值
    for Sample_name, values in Samples_info.items():
        sample_mean = sum(values) / len(values)
        mean_list.append((Sample_name, sample_mean))

    # Step 2: 计算 batch 全局均值
    mean_list_values = [v for _, v in mean_list]
    batch_mean = sum(mean_list_values) / len(mean_list_values)

    # Step 3: 构造 Samples 和 sample_diff_list
    sample_diff_list = []
    for Sample_name, sample_mean in mean_list:
        sample_diff = batch_mean - sample_mean
        abs_diff = abs(sample_diff)

        if abs_diff < 2:
            label = "aligned"
        elif 2 <= abs_diff <= 5:
            label = "mild_shift"
        else:
            label = "strong_shift"

        Samples[Sample_name] = {
            'sample_mean': sample_mean,
            'sample_diff': sample_diff,
            'label': label
        }
        sample_diff_list.append((Sample_name, abs_diff))



    group_summary = {}

    for sample_name, details in Samples.items():
        label = details['label']
        deviation = abs(details['sample_diff'])  # 用绝对偏差值统计

        if label not in group_summary:
            group_summary[label] = {'counts': 0, 'deviation_sum': 0}

        group_summary[label]['counts'] += 1
        group_summary[label]['deviation_sum'] += deviation

    # 计算平均偏差
    for label in group_summary:
        count = group_summary[label]['counts']
        total = group_summary[label]['deviation_sum']
        group_summary[label]['avg_deviation'] = total / count
        del group_summary[label]['deviation_sum']


    # Step 4: 排序并赋 rank
    sample_diff_list.sort(key=lambda x: x[1], reverse=True)
    rank = 1
    for Sample_name, _ in sample_diff_list:
        Samples[Sample_name]['rank'] = rank
        rank += 1

    # Step 5: 保存结果
    batch_info[batch] = {'Samples': Samples,
                        'group_summary': group_summary }

print(batch_info)
````

````python
def analyze_group_summary(Samples):
    group_summary = {}

    for sample_name, details in Samples.items():
        label = details['label']
        deviation = abs(details['sample_diff'])  # 用绝对偏差值统计

        if label not in group_summary:
            group_summary[label] = {'counts': 0, 'deviation_sum': 0}

        group_summary[label]['counts'] += 1
        group_summary[label]['deviation_sum'] += deviation
    return group_summary
````

````python
def analyz_rna_batch(rna_batches):
    batch_info = {}
    for Batch,Samples_info in rna_batches.items():
        Sample_details = {}
        for Sample,values in Samples_info.items():
            Sample_mean = sum(values) / len(values)
            Sample_diff_square = [(v - Sample_mean)**2 for v in values]
            Sample_var = sum(Sample_diff_square) / len(Sample_diff_square)
            Sample_std = round(Sample_var**0.5,3)
            Sample_CV = round(Sample_std / Sample_mean,3)
            if Sample_CV < 0.05 :
                label = "highly_consistent"
            elif  0.05 <= Sample_CV <0.15 :
                label = "moderate_consistent"
            else :
                label = "variable"
            Sample_details[Sample] = {'Sample_std':Sample_std,'Sample_mean':Sample_mean,
                                    'Sample_CV': Sample_CV,'label':label}
    
        CV_list = []
        for Sample , details in Sample_details.items():
            Sample_CV = details['Sample_CV']
            CV_list.append((Sample,Sample_CV))

        for i in range(len(CV_list)):
            for j in range(len(CV_list)-i-1):
                if CV_list[j][1] < CV_list[j+1][1]:
                    CV_list[j],CV_list[j+1] = CV_list[j],CV_list[j+1]
        
        rank = 1
        for Sample,_ in CV_list:
            Sample_details[Sample]['rank'] = rank 
            rank += 1
    
        batch_quality_summary = {}
        for Sample ,details in Sample_details.items():
            label = details['label']
            deviation = details['Sample_CV']
            if label not in batch_quality_summary:
                batch_quality_summary[label] = {'counts': 0,'deviation_sum':0}
            batch_quality_summary[label]['counts'] += 1
            batch_quality_summary[label]['deviation_sum'] += deviation

        # 计算平均偏差
        for label in batch_quality_summary:
            count = batch_quality_summary[label]['counts']
            total = batch_quality_summary[label]['deviation_sum']
            batch_quality_summary[label]['avg_deviation'] = total / count
            del batch_quality_summary[label]['deviation_sum']
        batch_info[Batch] = {
            "Samples": Sample_details,
            "GroupSummary": batch_quality_summary
        }
        return batch_info
````
