🧬 生信练习题：RNA-seq 样本表达量分析与分组统计
你有一个 RNA-seq 表达量数据结构 rna_batches，表示多个批次中不同样本的基因表达量（TPM）。请你编写程序完成以下任务：

```python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    }
}
```

---
🎯 要求：
计算每个样本的平均表达量（mean）和标准差（std）

根据平均表达量分组：

≥ 25："high_expression"

≥ 15："medium_expression"

≥ 5："low_expression"

否则："very_low"

在每个 batch 内按平均表达量从高到低排序并分配排名

统计每个 batch 中各表达等级的样本数量和平均表达量

找出每个 batch 中表达量最低的样本

---
✅ 输出结构示例：
```python
rna_info = {
    "Batch_1": {
        "samples": {
            "Sample_A": {"values": [...], "mean": ..., "std": ..., "group": ..., "rank": ...},
            ...
        },
        "group_summary": {
            "high_expression": {"count": ..., "avg_mean": ...},
            "low_expression": {...},
            ...
        },
        "lowest_sample": "Sample_B"
    },
    ...
}
```

````python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    }
}
rna_batch_info= {}
for batch,sample_info in rna_batches.items():
    Samples = {}
    group_summary = {}
    mean_list = []
    for sample_name,values in sample_info.items():
        mean = sum(values)/len(values)
        mean_list.append((sample_name,mean))
        if all(v >= 25 for v in values):
            quality = "high_expression"
        elif all(v >= 15 for v in values):
            quality = "medium_expression"
        elif all(v >= 10 for v in values):
            quality = "low_expression"
        else:
            quality = "very low"
        Samples[sample_name] = {'quality':quality,'mean':mean}
    
    for i in range(len(mean_list)):
        for j in range(len(mean_list)-i-1):
            if mean_list[j][1] < mean_list[j+1][1]:
                mean_list[j],mean_list[j+1] = mean_list[j+1],mean_list[j]
    rank = 1
    for sample_name,_ in mean_list:
        Samples[sample_name]['rank'] = rank
        rank += 1
    
    for sample_name,details in Samples.items():
        q = details['quality']
        mean = details['mean']
        if q not in group_summary:
            group_summary[q] = {'counts':0,'mean':mean}
        group_summary[q]['counts'] += 1
        avg_mean = mean/group_summary[q]['counts']
        group_summary[q]['avg_mean']=avg_mean
    rna_batch_info[batch]={'Samples':Samples,'group_summary':group_summary}
print(Samples) 
print(mean_list)
print(group_summary)
print(rna_batch_info)
````

````python
rna_batches = {
    "Batch_1": {
        "Sample_A": [12.5, 15.2, 13.8, 14.1],
        "Sample_B": [5.1, 4.8, 5.3, 5.0],
        "Sample_C": [20.0, 22.1, 21.5, 19.8],
        "Sample_D": [8.5, 9.0, 8.8, 9.2]
    },
    "Batch_2": {
        "Sample_E": [30.5, 32.1, 31.8, 33.0],
        "Sample_F": [10.0, 9.8, 10.2, 10.1],
        "Sample_G": [2.5, 2.8, 2.6, 2.7],
        "Sample_H": [18.0, 17.5, 18.2, 17.8]
    },
    "Batch_3": {
        "Sample_I": [25.0, 24.5, 25.5, 26.0],
        "Sample_J": [14.0, 13.5, 14.2, 13.8],
        "Sample_K": [6.0, 6.5, 6.2, 6.1],
        "Sample_L": [19.5, 20.0, 19.8, 20.2]
    },
    "Batch_4": {
        "Sample_M": [11.0, 11.5, 11.2, 11.3],
        "Sample_N": [3.0, 3.5, 3.2, 3.1],
        "Sample_O": [16.0, 16.5, 16.2, 16.1],
        "Sample_P": [27.0, 28.5, 27.8, 28.2]
    }
}

'''
🧠 任务目标：
对每个样本计算表达标准差（std）

根据 std 值打标签：

std < 0.3 → "stable"

0.3 ≤ std < 1.0 → "moderate"

std ≥ 1.0 → "unstable"

按 std 值降序排序并赋 rank

汇总每个标签的样本数和平均 std

封装为函数：analyze_expression_variability(batch_name, rna_batches)
✅ 练习要求：
使用 statistics.stdev() 或 numpy.std() 计算标准差

不引入全局变量，函数应独立运行

保持结构清晰，变量命名合理

可选：支持多个批次循环调用
'''

for batch,sample_info in rna_batches.items():
    Samples = {}
    for Sample_name,values in sample_info.items():
        mean = sum(values)/len(values)
        diff = [(v - mean)**2 for v in values]
        std = (sum(diff)/len(values))**0.5
        if std < 0.3:
            label = 'stable'
        elif 0.3 <= std < 1.0:
            label = 'moderate'
        else:
            label = 'unstable'
        Samples[Sample_name] = {'std':std,'label':label}
    std_list = []
    for Sample_name,details in Samples.items():
        std = details['std']
        std_list.append((Sample_name,std))
    for i in range(len(std_list)):
        for j in range(len(std_list)-i-1):
            if std_list[j][1] < std_list[j+1][1]:
                std_list[j],std_list[j+1] = std_list[j+1],std_list[j]
    rank = 1
    for Sample_name,_ in std_list:
        Samples[Sample_name]['rank'] = rank
        rank += 1
````
